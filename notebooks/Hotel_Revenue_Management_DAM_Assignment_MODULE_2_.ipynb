2.1 – Import Required Libraries

```python
# 2.1 – Import Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully.")
```

2.3 – Load the Dataset

```python
# 2.3 – Load Hotel Booking Dataset

df = pd.read_csv('hotel_bookings.csv')

print("Dataset loaded successfully.")
```

2.4 – Check Dataset Dimensions

```python
# 2.4 – Check Dataset Dimensions

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

2.5 – Check Dataset Information

```python
# 2.5 – Check Dataset Information

df.info()
```

2.6 – Identify Missing Values in Each Column

```python
# 2.6 – Identify Missing Values

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values[missing_values > 0].sort_values(ascending=False))
```

2.7 – Calculate Missing Value Percentage

```python
# 2.7 – Calculate Missing Value Percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Missing Percentage': missing_percentage
})

missing_summary = missing_summary[
    missing_summary['Missing Values'] > 0
]

missing_summary.sort_values(
    by='Missing Percentage',
    ascending=False
)
```

```python
# 2.8 – Visualise Missing Values

plt.figure(figsize=(14, 6))

sns.heatmap(df.isnull(), cbar=False, yticklabels=False)

plt.title("Missing Values in Hotel Booking Dataset")
plt.xlabel("Columns")
plt.ylabel("Booking Records")

plt.show()
```

2.9 – Identify Duplicate Records

```python
# 2.9 – Identify Duplicate Records

duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)
```

2.10 – Display Duplicate Records

```python
# 2.10 – Display Duplicate Records

duplicate_records = df[df.duplicated(keep=False)]

print("First 10 Duplicate Records:")

display(duplicate_records.head(10))
```

2.11 – Check Duplicate Records by Hotel Type

```python
# 2.11 – Check Duplicate Records by Hotel Type

duplicate_records = df[df.duplicated()]

duplicate_by_hotel = duplicate_records.groupby(
    'hotel'
).size().reset_index(name='Duplicate_Count')

display(duplicate_by_hotel)
```

2.12 – Calculate Unique Records

```python
# 2.12 – Calculate Unique Records

total_records = len(df)

duplicate_count = df.duplicated().sum()

unique_records = total_records - duplicate_count

print("Total Records:", total_records)
print("Duplicate Records:", duplicate_count)
print("Unique Records:", unique_records)
```

2.13 – Check Unique Values in Categorical Variables

```python
# 2.13 – Check Unique Values in Categorical Variables

categorical_columns = df.select_dtypes(
    include='object'
).columns

for column in categorical_columns:

    print("\nColumn:", column)

    print("Unique Values:")
    print(df[column].value_counts(dropna=False))
```

2.14 – Identify Inconsistent Entries

```python
# 2.14 – Identify Inconsistent Entries

for column in categorical_columns:

    print("\nColumn:", column)

    values = df[column].dropna().astype(str)

    leading_trailing_spaces = values[
        values != values.str.strip()
    ]

    print(
        "Entries with Extra Spaces:",
        len(leading_trailing_spaces)
    )

    print(
        "Unique Categories:",
        df[column].nunique()
    )
```

2.15 – Check Missing Values in Important Columns

```python
# 2.15 – Check Missing Values in Important Columns

important_columns = [
    'adr',
    'hotel',
    'country',
    'agent',
    'company',
    'children',
    'market_segment',
    'customer_type'
]

missing_important = df[important_columns].isnull().sum()

print("Missing Values in Important Columns:")

display(missing_important.to_frame(
    name='Missing_Count'
))
```

2.16 – Identify Outliers Using the IQR Method

```python
# 2.16 – Identify Outliers Using IQR Method

selected_numeric_columns = [
    'adr',
    'lead_time',
    'adults',
    'children',
    'babies',
    'stays_in_weekend_nights',
    'stays_in_week_nights',
    'previous_cancellations',
    'previous_bookings_not_canceled',
    'booking_changes',
    'days_in_waiting_list',
    'required_car_parking_spaces',
    'total_of_special_requests'
]

outlier_results = []

for column in selected_numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - (1.5 * IQR)
    upper_limit = Q3 + (1.5 * IQR)

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ][column].count()

    outlier_results.append({
        'Variable': column,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Limit': lower_limit,
        'Upper Limit': upper_limit,
        'Outlier Count': outliers
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary.round(2))
```

2.17 – Visualise Outliers Using Box Plots

```python
# 2.17 – Visualise Outliers Using Box Plots

for column in selected_numeric_columns:

    plt.figure(figsize=(10, 4))

    sns.boxplot(x=df[column])

    plt.title("Outlier Detection in " + column)
    plt.xlabel(column)

    plt.tight_layout()
    plt.show()
```

2.18 – Identify Negative and Zero ADR Values

```python
# 2.18 – Check Negative and Zero ADR Values

negative_adr = (df['adr'] < 0).sum()

zero_adr = (df['adr'] == 0).sum()

positive_adr = (df['adr'] > 0).sum()

print("Negative ADR Records:", negative_adr)

print("Zero ADR Records:", zero_adr)

print("Positive ADR Records:", positive_adr)
```

2.19 – Analyse Average ADR by Hotel Type

```python
# 2.19 – Average ADR by Hotel Type

hotel_adr_analysis = df.groupby('hotel').agg(
    Total_Bookings=('adr', 'count'),
    Average_ADR=('adr', 'mean'),
    Median_ADR=('adr', 'median'),
    Minimum_ADR=('adr', 'min'),
    Maximum_ADR=('adr', 'max')
).round(2)

display(hotel_adr_analysis)
```

2.20 – Analyse Bookings by Market Segment

```python
# 2.20 – Bookings by Market Segment

market_segment_analysis = df.groupby(
    'market_segment'
).agg(
    Total_Bookings=('adr', 'count'),
    Average_ADR=('adr', 'mean'),
    Median_ADR=('adr', 'median')
).round(2)

display(market_segment_analysis)
```

2.21 – Analyse Bookings by Customer Type

```python
# 2.21 – Bookings by Customer Type

customer_type_analysis = df.groupby(
    'customer_type'
).agg(
    Total_Bookings=('adr', 'count'),
    Average_ADR=('adr', 'mean'),
    Median_ADR=('adr', 'median')
).round(2)

display(customer_type_analysis)
```

2.22 – Analyse Average ADR by Arrival Month

```python
# 2.22 – Average ADR by Arrival Month

month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

monthly_adr_analysis = df.groupby(
    'arrival_date_month'
)['adr'].mean().reindex(month_order).round(2)

display(monthly_adr_analysis.to_frame(
    name='Average_ADR'
))
```

2.23 – Analyse Average ADR by Meal Type

```python
# 2.23 – Average ADR by Meal Type

meal_analysis = df.groupby('meal').agg(
    Total_Bookings=('adr', 'count'),
    Average_ADR=('adr', 'mean')
).round(2)

display(meal_analysis)
```

2.24 – Analyse Average ADR by Reserved Room Type

```python
# 2.24 – Average ADR by Reserved Room Type

room_type_analysis = df.groupby(
    'reserved_room_type'
).agg(
    Total_Bookings=('adr', 'count'),
    Average_ADR=('adr', 'mean'),
    Median_ADR=('adr', 'median')
).round(2)

display(room_type_analysis)
```

2.25 – Compare Hotel Type and Market Segment

```python
# 2.25 – Hotel Type and Market Segment Analysis

hotel_segment_analysis = df.pivot_table(
    index='market_segment',
    columns='hotel',
    values='adr',
    aggfunc='mean'
).round(2)

display(hotel_segment_analysis)
```

**Data Cleaning**

2.27 – Create a Copy of the Original Dataset

```python
# 2.27 – Create a Copy of Original Dataset

df_clean = df.copy()

print("Original Dataset Shape:", df.shape)
print("Working Dataset Shape:", df_clean.shape)
```

2.28 – Remove Exact Duplicate Records

```python
# 2.28 – Remove Exact Duplicate Records

df_clean = df_clean.drop_duplicates()

print("Dataset Shape After Removing Duplicates:", df_clean.shape)

print("Remaining Duplicate Records:", df_clean.duplicated().sum())
```

2.29 – Handle Missing Values in Children Column

```python
# 2.29 – Handle Missing Values in Children Column

df_clean['children'] = df_clean['children'].fillna(0)

print("Missing Values in Children:",
      df_clean['children'].isnull().sum())
```

2.30 – Handle Missing Values in Country Column

```python
# 2.30 – Handle Missing Values in Country Column

df_clean['country'] = df_clean['country'].fillna('Unknown')

print("Missing Values in Country:",
      df_clean['country'].isnull().sum())
```

2.31 – Check Missing Values After Cleaning

```python
# 2.31 – Check Missing Values After Cleaning

missing_after_cleaning = df_clean.isnull().sum()

print("Missing Values After Cleaning:")

display(
    missing_after_cleaning[
        missing_after_cleaning > 0
    ].sort_values(ascending=False)
)
```

2.32 – Check Negative ADR Values

```python
# 2.32 – Check Negative ADR Values

negative_adr_records = df_clean[df_clean['adr'] < 0]

print("Negative ADR Records:",
      len(negative_adr_records))

display(negative_adr_records)
```

2.33 – Prepare Dataset for Regression

```python
# 2.33 – Prepare Dataset for Regression

df_model = df_clean[df_clean['adr'] >= 0].copy()

print("Clean Dataset Shape:", df_model.shape)

print("Negative ADR Records Remaining:",
      (df_model['adr'] < 0).sum())

print("Zero ADR Records:",
      (df_model['adr'] == 0).sum())
```

2.34 – Final Data Cleaning Summary

```python
# 2.34 – Final Data Cleaning Summary

print("FINAL DATA CLEANING SUMMARY")

print("\nOriginal Dataset Rows:", len(df))

print("Rows After Duplicate Removal:", len(df_clean))

print("Rows Prepared for Regression:", len(df_model))

print("Missing Children Values:",
      df_model['children'].isnull().sum())

print("Missing Country Values:",
      df_model['country'].isnull().sum())

print("Negative ADR Values:",
      (df_model['adr'] < 0).sum())

print("Remaining Duplicate Records:",
      df_model.duplicated().sum())
```
