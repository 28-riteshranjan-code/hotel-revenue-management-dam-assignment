2.16 – Identify Outliers Using the IQR Method

```python
# 2.16 – Identify Outliers Using IQR Method

selected_numeric_columns = [
    'adr',
    'lead_time',
    'adults',
    'children',
    'babies',
    'stays_in_weekend_nights',
    'stays_in_week_nights',
    'previous_cancellations',
    'previous_bookings_not_canceled',
    'booking_changes',
    'days_in_waiting_list',
    'required_car_parking_spaces',
    'total_of_special_requests'
]

outlier_results = []

for column in selected_numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - (1.5 * IQR)
    upper_limit = Q3 + (1.5 * IQR)

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ][column].count()

    outlier_results.append({
        'Variable': column,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Limit': lower_limit,
        'Upper Limit': upper_limit,
        'Outlier Count': outliers
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary.round(2))
```

3.25 – Prepare Dataset for Regression

```python
# 3.25 – Prepare Dataset for Regression

regression_columns = [
    'adr',
    'lead_time',
    'adults',
    'children',
    'babies',
    'stays_in_weekend_nights',
    'stays_in_week_nights',
    'total_of_special_requests',
    'is_repeated_guest',
    'hotel',
    'market_segment',
    'distribution_channel',
    'customer_type',
    'meal',
    'reserved_room_type',
    'arrival_date_month',
    'arrival_date_year'
]

regression_data = df_model[regression_columns].copy()

# Handle any remaining missing or infinite values
regression_data = regression_data.replace(
    [np.inf, -np.inf], np.nan
).dropna()

print("Regression Dataset Shape:", regression_data.shape)

print("\nDependent Variable: adr")

print("\nNumerical Predictors:")
print(regression_data.select_dtypes(include=np.number).columns.tolist())

print("\nCategorical Predictors:")
print(regression_data.select_dtypes(include='object').columns.tolist())
```

```python
# 3.26 – Split Dataset into Training and Testing Data

from sklearn.model_selection import train_test_split

train_data, test_data = train_test_split(
    regression_data,
    test_size=0.20,
    random_state=42
)

print("Training Records:", len(train_data))

print("Testing Records:", len(test_data))
```

3.27 – Build Multiple Linear Regression Model Without Interaction

```python
# 3.27 – Multiple Linear Regression Without Interaction

formula1 = """
adr ~ lead_time
+ adults
+ children
+ babies
+ stays_in_weekend_nights
+ stays_in_week_nights
+ total_of_special_requests
+ is_repeated_guest
+ C(hotel, Treatment(reference='City Hotel'))
+ C(market_segment, Treatment(reference='Online TA'))
+ C(distribution_channel, Treatment(reference='TA/TO'))
+ C(customer_type, Treatment(reference='Transient'))
+ C(meal, Treatment(reference='BB'))
+ C(reserved_room_type, Treatment(reference='A'))
+ C(arrival_date_month, Treatment(reference='January'))
+ C(arrival_date_year, Treatment(reference=2015))
"""

model1 = smf.ols(
    formula=formula1,
    data=train_data
).fit()

print(model1.summary())
```

**2. Multiple Linear Regression With Interaction**

3.28 – Build Regression Model With Interaction

```python
# 3.28 – Multiple Linear Regression With Interaction

formula2 = formula1 + """
+ C(hotel, Treatment(reference='City Hotel'))
: C(arrival_date_month, Treatment(reference='January'))
"""

model2 = smf.ols(
    formula=formula2,
    data=train_data
).fit()

print(model2.summary())
```

3.29 – Display Interaction Coefficients

```python
# 3.29 – Display Interaction Coefficients

interaction_coefficients = pd.DataFrame({
    'Coefficient': model2.params,
    'P_Value': model2.pvalues
})

interaction_coefficients = interaction_coefficients[
    interaction_coefficients.index.str.contains(':')
]

display(interaction_coefficients.round(4))
```

3.30 – Compare Both Regression Models

```python
# 3.30 – Compare Both Regression Models

from sklearn.metrics import mean_squared_error, mean_absolute_error

# Predictions on the same testing data
prediction1 = model1.predict(test_data)
prediction2 = model2.predict(test_data)

# Model comparison
comparison = pd.DataFrame({
    'Measure': [
        'R-Squared',
        'Adjusted R-Squared',
        'AIC',
        'BIC',
        'F-Statistic',
        'F-Statistic P-Value',
        'Test RMSE',
        'Test MAE',
        'Significant Predictors'
    ],

    'Model 1 Without Interaction': [
        model1.rsquared,
        model1.rsquared_adj,
        model1.aic,
        model1.bic,
        model1.fvalue,
        model1.f_pvalue,
        np.sqrt(mean_squared_error(test_data['adr'], prediction1)),
        mean_absolute_error(test_data['adr'], prediction1),
        (model1.pvalues.drop('Intercept', errors='ignore') < 0.05).sum()
    ],

    'Model 2 With Interaction': [
        model2.rsquared,
        model2.rsquared_adj,
        model2.aic,
        model2.bic,
        model2.fvalue,
        model2.f_pvalue,
        np.sqrt(mean_squared_error(test_data['adr'], prediction2)),
        mean_absolute_error(test_data['adr'], prediction2),
        (model2.pvalues.drop('Intercept', errors='ignore') < 0.05).sum()
    ]
})

display(comparison.round(4))
```

3.31 – Identify the Preferred Regression Model

```python
# 3.31 – Identify Preferred Regression Model

rmse1 = np.sqrt(
    mean_squared_error(test_data['adr'], prediction1)
)

rmse2 = np.sqrt(
    mean_squared_error(test_data['adr'], prediction2)
)

print("Model 1 Test RMSE:", round(rmse1, 2))
print("Model 2 Test RMSE:", round(rmse2, 2))

if model2.aic < model1.aic and rmse2 < rmse1:
    print("\nModel 2 is preferred based on AIC and test RMSE.")
elif model1.aic < model2.aic and rmse1 < rmse2:
    print("\nModel 1 is preferred based on AIC and test RMSE.")
else:
    print("\nThe models show mixed results. Further comparison is required.")
```
